# DooDL: outline an object once, track it through the video

1. Choose **Runtime → Change runtime type → T4 GPU** (once).
2. Choose **Runtime → Run all**. The app appears below in about 3 minutes.
3. Click **Use a sample clip** to try it right away, or upload your own video (up to 30 s).

```
#@title Start DooDL { display-mode: "form" }
REPO = "Ahmed-ssam/DooDL-DrawToTrack"  #@param {type:"string"}
MODEL = "tiny"  #@param ["tiny", "small", "base_plus", "large"]
SHARE_PUBLIC_LINK = False  #@param {type:"boolean"}

import subprocess, sys

def pip(*packages):
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", *packages], check=True)

has_gpu = subprocess.run(["nvidia-smi"], capture_output=True).returncode == 0
if has_gpu:
    print("GPU found. Installing SAM 2.1 (about 2 minutes)...")
    pip("git+https://github.com/facebookresearch/sam2.git")
else:
    print("No GPU in this runtime, so DooDL will use its basic CPU tracker.")
    print("For SAM 2.1 choose Runtime -> Change runtime type -> T4 GPU, then run again.")
print("Installing DooDL...")
pip(f"git+https://github.com/{REPO}.git")

from doodl import colab
app = colab.launch(backend="sam2" if has_gpu else "classical", sam2_model=MODEL, share=SHARE_PUBLIC_LINK)
```

**Tips.** Keep clips under 30 s. If you see an out-of-memory error, use `tiny` or a shorter clip. To stop the app, run `app.stop()`. Turn on *SHARE_PUBLIC_LINK* to get a link you can open on another device; it stops working when this notebook stops.
